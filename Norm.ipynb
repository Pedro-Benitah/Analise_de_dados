```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# Importação dos dados
df=pd.read_csv('spotify_churn_dataset.csv')
df.head(2)
```

```python
# Verificar valores únicos nas colunas categóricas para identificar possíveis incorreções
print("Valores únicos na coluna gender:")
print(df['gender'].unique())

print("\nValores únicos na coluna country:")
print(df['country'].unique())

print("\nValores únicos na coluna subscription_type:")
print(df['subscription_type'].unique())

print("\nValores únicos na coluna device_type:")
print(df['device_type'].unique())
```

```python
print("Quantidade de valores nulos na coluna user_id:")
print(df['user_id'].isnull().sum())
print("Quantidade de valores nulos na coluna gender:")
print(df['gender'].isnull().sum())
print("Quantidade de valores nulos na coluna age:")
print(df['age'].isnull().sum())
print("Quantidade de valores nulos na coluna country:")
print(df['country'].isnull().sum())
print("Quantidade de valores nulos na coluna subscription_type:")
print(df['subscription_type'].isnull().sum())
print("Quantidade de valores nulos na coluna listening_time:")
print(df['listening_time'].isnull().sum())
print("Quantidade de valores nulos na coluna songs_played_per_day:")
print(df['songs_played_per_day'].isnull().sum())
print("Quantidade de valores nulos na coluna skip_rate:")
print(df['skip_rate'].isnull().sum())
print("Quantidade de valores nulos na coluna device_type:")
print(df['device_type'].isnull().sum())
print("Quantidade de valores nulos na coluna ads_listened_per_week:")
print(df['ads_listened_per_week'].isnull().sum())
print("Quantidade de valores nulos na coluna offline_listening:")
print(df['offline_listening'].isnull().sum())
print("Quantidade de valores nulos na coluna is_churned:")
print(df['is_churned'].isnull().sum())
```

```python
import matplotlib.pyplot as plt

cols_numericas = ["age", "listening_time", "songs_played_per_day",
                  "skip_rate", "ads_listened_per_week"]

df[cols_numericas].plot(kind="box", subplots=True, layout=(2,3), figsize=(12,6), sharex=False, sharey=False)
plt.tight_layout()
plt.show()
```

```python
df[cols_numericas].hist(bins=30, figsize=(12,6))
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12,6))
for i, col in enumerate(cols_numericas, 1):
    plt.subplot(2,3,i)
    sns.violinplot(y=df[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
# Estatísticas descritivas
display(df.describe(include="all").transpose())

# Distribuição das variáveis numéricas
df.hist(bins=30, figsize=(12, 8))
plt.suptitle("Distribuição das variáveis numéricas", fontsize=14)
plt.tight_layout()
plt.show()

# Boxplots para verificar possíveis outliers
cols_numericas = ["age", "listening_time", "songs_played_per_day",
                  "skip_rate", "ads_listened_per_week"]

plt.figure(figsize=(12,6))
df[cols_numericas].plot(kind="box", subplots=True, layout=(2,3),
                        figsize=(12,6), sharex=False, sharey=False)
plt.suptitle("Boxplots das variáveis numéricas", fontsize=14)
plt.tight_layout()
plt.show()

# Distribuição do churn (alvo)
plt.figure(figsize=(5,4))
sns.countplot(x="is_churned", data=df)
plt.title("Distribuição do Status de Cancelamento")
plt.xlabel("Status (0 = Fidelizado, 1 = Cancelado)")
plt.ylabel("Quantidade")
plt.show()
```
